# Data analysis

Reads the synthetic ticket file written by `data-eval run`.


```python
import json
from collections import Counter
from pathlib import Path

rows = [
    json.loads(line)
    for line in Path('../data/raw/tickets.jsonl').read_text(encoding='utf-8').splitlines()
    if line.strip()
]
print('tickets', len(rows))
print('classes', dict(sorted(Counter(row['gold_type'] for row in rows).items())))
print('splits', dict(sorted(Counter(row['split'] for row in rows).items())))
print('difficulties', dict(sorted(Counter(row['difficulty'] for row in rows).items())))
```
