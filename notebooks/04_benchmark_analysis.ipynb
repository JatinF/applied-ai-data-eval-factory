# Benchmark analysis

Compares v1 and v2 manifests. Item ids should match aside from the version prefix.


```python
import json
from pathlib import Path

def load(name: str) -> list[dict]:
    path = Path('../data/benchmark') / name
    return [json.loads(line) for line in path.read_text(encoding='utf-8').splitlines() if line.strip()]

v1 = load('benchmark_v1.jsonl')
v2 = load('benchmark_v2.jsonl')
manifest = json.loads(Path('../data/benchmark/manifest.json').read_text(encoding='utf-8'))
print('v1', len(v1), 'v2', len(v2))
print('same tickets', {row['ticket_id'] for row in v1} == {row['ticket_id'] for row in v2})
for version in manifest['versions']:
    print(version['version'], 'tasks', version['task_counts'])
    print(version['version'], 'difficulty', version['difficulty_counts'])
    print(version['version'], 'slices', version['slice_counts'])
```
