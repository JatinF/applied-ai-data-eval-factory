# Synthetic data

Counts provenance fields on `data/synthetic/synthetic.jsonl`.


```python
import json
from collections import Counter
from pathlib import Path

rows = [
    json.loads(line)
    for line in Path('../data/synthetic/synthetic.jsonl').read_text(encoding='utf-8').splitlines()
    if line.strip()
]
print('rows', len(rows))
print('strategy', dict(sorted(Counter(row['generation_strategy'] for row in rows).items())))
print('review', dict(sorted(Counter(row['review_status'] for row in rows).items())))
print('generator', sorted({row['generator'] for row in rows}))
print('prompt_version', sorted({row['prompt_version'] for row in rows}))
```
