# Weak supervision

Prints coverage, overlap, conflict, and per-function accuracy from the measured report.


```python
import json
from pathlib import Path

report = json.loads(Path('../reports/baseline_metrics.json').read_text(encoding='utf-8'))
labeling = report['v1']['labeling']
print('backend', labeling['backend'], 'used_gold_to_fit', labeling['used_gold_to_fit'])
print('coverage', labeling['coverage'], 'overlap', labeling['overlap'], 'conflict', labeling['conflict'])
for item in labeling['labeling_functions']:
    print(
        f"{item['name']}: coverage={item['coverage']:.4f} "
        f"estimated={item['estimated_accuracy']:.4f} gold={item['gold_accuracy']}"
    )
```
