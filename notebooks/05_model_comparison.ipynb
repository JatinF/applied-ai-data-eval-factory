# Model comparison

Prints the measured scorecard. Hosted LLM and LoRA stay `not-run`.


```python
import json
from pathlib import Path

report = json.loads(Path('../reports/baseline_metrics.json').read_text(encoding='utf-8'))
print('hosted_llm', report['hosted_llm'], 'lora', report['lora'])
print('kappa', report['cohen_kappa'], 'adjudicated', report['adjudicated_count'])
for block in (report['v1'], report['v2']):
    print('iteration', block['iteration'])
    for score in block['scores']:
        print(
            score['name'],
            'f1', round(score['f1_macro'], 4),
            'grounded', score['groundedness'],
            'recall@5', score['recall_at_5'],
            'p95_ms', round(score['p95_ms'], 4),
            'unsafe', score['unsafe_action_rate'],
            'tokens', score['tokens'],
        )
```
